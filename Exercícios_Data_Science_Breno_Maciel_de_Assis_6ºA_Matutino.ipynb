#### **04. Gere um array de 100 números inteiros aleatórios e encontre a mediana e a moda dos valores.**

```python
# Importo a biblioteca 'numpy', apelidando de 'np'.
import numpy as np

# Defino um array aleatório de 0 a 50 com 100 elementos.
array = np.random.randint(0, 51, size = 100)

# Calculo a mediana usando o recurso 'median', do numpy.
median = np.median(array)

# O módulo 'np.unique()' do numpy identifica todos os números distintos presentes no array.
# O parâmetro 'return_counts = True' faz gerar um array com a frequência que cada número apareceu.
values, count = np.unique(array, return_counts = True)

# A função 'np.argmax()' localiza a posição do valor mais alto dentro do array de contagens (o maior número será o mais frequente, ou seja, a moda)
mode_index = np.argmax(count)
# Dentro da variável 'mode', insiro o elemento que está no índice da moda
mode = values[mode_index]
# Dentro da variável 'frequency', insiro o elemento da contagem que está guardado no índice da moda
frequency = count[mode_index]

# Retorno o array, mediana, moda e sua frequência para o usuário
print(f"Array aleatório gerado:\n{array}")
print(f"\nMediana: {median}")
print(f"Moda: {mode} (Frequência: {frequency}x)")
```

#####**Explicação:**
Primeiro, eu importo a biblioteca 'numpy' como 'np'.

Então, defino um array gerando 100 números inteiros aleatórios que variam de 0 a 50 utilizando a função 'np.random.randint()' e calculo a mediana teórica do conjunto de dados com a função 'np.median()'.

Depois, utilizo a função np.unique() com o parâmetro opcional return_counts = True para mapear todos os números distintos presentes na distribuição juntamente com suas frequências correspondentes, armazenando-os ordenadamente nas variáveis 'values' e 'count'.

Então, para encontrar a moda do conjunto, localizo o índice do valor mais alto dentro do array de contagens por meio da instrução np.argmax(), extraio o elemento do conjunto de valores que representa a moda combinando a indexação com o índice achado, e também pego o número absoluto de ocorrências que esse valor registrou.

No fim, retorno os resultados das tendências exibindo as saídas formatadas com o array aleatório completo, a mediana, e o valor da moda junto com sua respectiva frequência estatística.

#### **09. Crie um array com 10 números aleatórios e normalize seus valores entre 0 e 1.**

```python
# Importo a biblioteca 'numpy', apelidando de 'np'.
import numpy as np

# Gero um array com 10 números aleatórios de uma distribuição normal, e multiplico por 10 para resultar em valores mais espalhados fora do intervaloo
original_array = np.random.randn(10) * 10

# Determino o valor mínimo e máximo do array
min_value = np.min(original_array)
max_value = np.max(original_array)

# Usando a técnica Min-Max Scaling, eu subtraio o menor valor do array de cada elemento e divido o resultado pela diferença entre o maior e o menor valor. (Como as operações do NumPy são vetorizadas, o 'min_value' é subtraído automaticamente de todos os elementos quando subtraído por 'original_array', e o mesmo conceito para o 'max_value')
normalized_array = (original_array - min_value) / (max_value - min_value)

# Retorno os resultados para o usuário
print(f"Array original: {original_array}")
print(f"\nMaior valor: {max_value}\nMenor valor: {min_value}")
print(f"\nArray normalizado: {normalized_array}")
```

#####**Explicação:**
De início, eu importo a biblioteca 'numpy' apelidando de 'np'.

Então, crio a variável chamada 'array' que recebe 10 números inteiros aleatórios entre 1 e 100 gerados de forma direta pela função np.random.randint().

Depois, utilizo as funções np.min() e np.max() para capturar as coordenadas do menor e do maior elemento e aplico a fórmula matemática de normalização simples para converter todos os valores do conjunto de dados para uma escala entre 0 e 1, armazenando na variável 'array_normalized'.

Só então retorno o array com as posições originais e a sequência dos novos valores normalizados no terminal.

#### **12. Verifique quais colunas de um DataFrame contém valores duplicados.**

```python
# Importo a biblioteca 'pandas', a apelidando de 'pd'
import pandas as pd

# Declaro o dict 'data' com os nomes, preços e quantidades dos produtos
data = {
    'Produto': ['Arroz', 'Azeite', 'Arroz', 'Kinder Ovo', 'Mistério'],
    'Preço': [29.90, 12.90, 12.90, 149.90, None],
    'Quantidade': [5, 6, 7, 8, None]
}

# Defino o dict 'data' em um DataFrame
df = pd.DataFrame(data)

# Definindo na variável 'columns_verification':
# O '.apply()' aplica funções como um loop
# O 'lambda' é um determinante de funções compactas e não nomeadas
# O '.duplicated()' é uma função que analisa os elementos da coluna e devolve 'True's pra cada valor que já tenha aparecido e 'Falses's se for a primeira vez que o valor aparece
# O '.any()' verifica se existe pelo menos um 'True' na lista, e se houver, retorna 'True', e caso todos os valores sejam únicos, ele retorna 'False'
columns_verification = df.apply(lambda column: column.duplicated().any())

# Defino na variável 'duplicated_columns' as colunas que contém os valores duplicados
# 'columns_verification[columns_verification]' é a indexação booleana na qual o pandas analisa as linhas e mantém os itens que os valores associados são 'True'
# O '.index' retorna os índices (que são os nomes das colunas originais)
# O '.tolist()' transforma o objeto em uma lista comum
duplicated_columns = columns_verification[columns_verification].index.tolist()

# Retorno o DataFrame, uma tabela que mostra as colunas que possuem dados duplicados (com True e False), e uma lista com os nomes das colunas que possuem valores duplicados
print(f"DataFrame analisado:\n{df}")
print(f"\nVerificação de valores duplicados por coluna:\n{columns_verification}")
print(f"\nColunas que contém valores duplicados:\n{duplicated_columns}")
```

#### **14. Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.**

```python
# Importo a biblioteca 'pandas', a apelidando de 'pd'
import pandas as pd

# Declaro o dict 'data1' com os ID's, nomes, categorias, preços, e quantidades dos produtos
data1 = {
    'ID': [1, 2, 3, 4, 5, 6],
    'Produto': ['Arroz', 'Azeite', 'Feijão', 'Kinder Ovo', 'Chocolate', 'Ketchup'],
    'Categoria': ['Alimentos', 'Condimentos', 'Alimentos', 'Doces', 'Doces', 'Condimentos'],
    'Preço': [29.90, 14.90, 12.90, 149.90, 99.90, 9.90],
    'Quantidade': [5, 6, 7, 8, 9, 3]
}

# Defino o dict 'data1' em um DataFrame
df1 = pd.DataFrame(data1)

# Declaro o dict 'data2' com os nomes, categorias, preços e quantidades dos produtos
data2 = {
    'ID': [1, 2, 3, 4, 5, 6],
    'Fornecedor': ['Fornecedor A', 'Fornecedor B', 'Fornecedor C', 'Fornecedor D', 'Fornecedor E', 'Fornecedor F']
}

# Defino o dict 'data2' em um DataFrame
df2 = pd.DataFrame(data2)

# Defino o 'inner join' na variável 'df_inner', usando o módulo '.merge()' que tem como parâmetros as duas tabelas, o elemento de unificação (ID), e o método 'inner' (que retorna os registros cujos ID's existem em ambos os DataFrames)
df_inner = pd.merge(df1, df2, on='ID', how='inner')
# Defino o 'left join' na variável 'df_left', usando o módulo '.merge()' que tem como parâmetros as duas tabelas, o elemento de unificação (ID), e o método 'left' (que mantém todos os dados do DataFrame 1, e caso não houvesse fornecedores no DataFrame 2, ele preencheria como NaN)
df_left = pd.merge(df1, df2, on='ID', how='left')

# Retorno os DataFrames para o usuário
print(f"DataFrame 1:\n{df1}")
print(f"\nDataFrame 2:\n{df2}")
# Retorno os resultados dos joins para o usuário
print(f"\nResultado do Inner Join (intersecção dos ID's):\n{df_inner}")
print(f"\nResultado do Left Join (primeira tabela preservada):\n{df_left}")
```

#### **19. Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.**

```python
# Importo a biblioteca 'matplotlib', a apelidando de 'plt'
import matplotlib.pyplot as plt
# Importo a biblioteca 'numpy', a apelidando de 'np'
import numpy as np

# Determino os 100 pontos de 0 a 10 homogeneamente espaçados
x = np.linspace(0, 10, 100)

# Determino a função da série A como Seno de X
a = np.sin(x)
# Determino a função da série B como Cosseno de X
b = np.cos(x)

# Ploto a série A:
# Eixo X sendo "x"
# Eixo Y sendo "a"
# Legenda sendo "Seno", e a cor da linha sendo laranja escuro
plt.plot(x, a, label="Seno", color="darkorange")

# Ploto a série B:
# Eixo X sendo "x"
# Eixo Y sendo "b"
# Legenda sendo "Cosseno", e a cor da linha sendo azul
plt.plot(x, b, label="Cosseno", color="skyblue")

# Personalização do gráfico:
# Título
plt.title("Gráfico de Comparação da Série A e B")
# Legenda do Eixo X
plt.xlabel("Eixo X")
# Legenda do Eixo Y
plt.ylabel("Eixo Y")
# Definição da legenda automática
plt.legend()

# Retorno o gráfico para o usuário
plt.show()
```

#### **20. Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x².**

```python
# Importo a biblioteca 'matplotlib', a apelidando de 'plt'
import matplotlib.pyplot as plt
# Importo a biblioteca 'numpy', a apelidando de 'np'
import numpy as np

# Gero os valores de x, sendo 100 pontos, de um intervalo de -10 a 10, homogeneamente separados
x = np.linspace(-10, 10, 100)
y  = x ** 2

# Ploto o gráfico:
# Eixo X definido por "x"
# Eixo Y definido por "y"
# Cor, grossura e legenda também definidas, respectivamente
plt.plot(x, y, color="skyblue", linewidth=2, label="y = x²")

# Defino os espaçamentos requisitados
x_spacing = np.linspace(-10, 10, 11)
y_spacing = np.linspace(0, 100, 11)

# Determino em quais coordenadas dos eixos as marcações dos números serão colocadas
plt.xticks(x_spacing)
plt.yticks(y_spacing)

# Personalização do gráfico:
# Título
plt.title("Gráfico de y = x² (com Grade Personalizada)")
# Legenda do Eixo X
plt.xlabel("Eixo X")
# Legenda do Eixo Y
plt.ylabel("Eixo Y")
# Definição da grade, cor, estilo, grossura e opacidade, respectivamente
plt.grid(True, color="gray", linestyle="--", linewidth=0.5, alpha=0.7)

# Retorno o gráfico para o usuário
plt.show()
```
